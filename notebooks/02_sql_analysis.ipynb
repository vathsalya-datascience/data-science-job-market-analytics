```python
conn=sqlite3.connect("/Users/Apple/OneDrive/Data science job market analytics/job_market.db")
```

```python
df.to_sql(
    "jobs",
    conn,
    if_exists="replace",
    index=False
)
```

```python
pd.read_sql(
    "SELECT COUNT(*) AS total_jobs FROM jobs;",
    conn
)
```

```python
query = """
SELECT
    ROUND(AVG(salary_in_usd), 2) AS average_salary,
    MIN(salary_in_usd) AS minimum_salary,
    MAX(salary_in_usd) AS maximum_salary
FROM jobs;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    job_title,
    COUNT(*) AS job_count
FROM jobs
GROUP BY job_title
ORDER BY job_count DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    experience_level_name,
    COUNT(*) AS job_count,
    ROUND(AVG(salary_in_usd), 2) AS average_salary
FROM jobs
GROUP BY experience_level_name
ORDER BY average_salary;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    work_arrangement,
    COUNT(*) AS job_count,
    ROUND(
        COUNT(*) * 100.0 / (SELECT COUNT(*) FROM jobs),
        2
    ) AS percentage
FROM jobs
GROUP BY work_arrangement
ORDER BY job_count DESC;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    company_size_name,
    COUNT(*) AS job_count,
    ROUND(
        COUNT(*) * 100.0 / (SELECT COUNT(*) FROM jobs),
        2
    ) AS percentage
FROM jobs
GROUP BY company_size_name
ORDER BY job_count DESC;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    company_location,
    COUNT(*) AS job_count,
    ROUND(AVG(salary_in_usd), 2) AS average_salary
FROM jobs
GROUP BY company_location
ORDER BY job_count DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    job_title,
    COUNT(*) AS job_count,
    ROUND(AVG(salary_in_usd), 2) AS average_salary
FROM jobs
GROUP BY job_title
HAVING COUNT(*) >= 50
ORDER BY average_salary DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    salary_category,
    COUNT(*) AS job_count,
    ROUND(AVG(salary_in_usd), 2) AS average_salary,
    ROUND(
        COUNT(*) * 100.0 / (SELECT COUNT(*) FROM jobs),
        2
    ) AS percentage
FROM jobs
GROUP BY salary_category
ORDER BY average_salary;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    work_year,
    COUNT(*) AS job_count,
    ROUND(AVG(salary_in_usd), 2) AS average_salary,
    ROUND(MIN(salary_in_usd), 2) AS minimum_salary,
    ROUND(MAX(salary_in_usd), 2) AS maximum_salary
FROM jobs
GROUP BY work_year
ORDER BY work_year;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    experience_level_name,
    job_title,
    COUNT(*) AS job_count,
    ROUND(AVG(salary_in_usd), 2) AS average_salary
FROM jobs
GROUP BY experience_level_name, job_title
HAVING COUNT(*) >= 50
ORDER BY experience_level_name, average_salary DESC;
"""

pd.read_sql(query, conn).head(20)
```
